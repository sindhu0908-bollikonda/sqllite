```python
import sqlite3

# Connect to the database (or create a new one if it doesn't exist)
conn = sqlite3.connect('your_database.db')
# Replace 'your_database.db' with your database file

# Create a cursor object
cursor = conn.cursor()

# Create the sales_data table if it doesn't exist
cursor.execute("""
CREATE TABLE IF NOT EXISTS sales_data (
    "Order ID" TEXT,
    "Order Date" TEXT,
    turnover REAL
)
""")

# Insert sample data (replace with your actual data)
sample_data = [
    ('Order1', '2019-01-15', 100.50),
    ('Order2', '2019-02-20', 250.75),
    ('Order3', '2019-03-10', 125.25),
    # ... more data
]
cursor.executemany("INSERT INTO sales_data VALUES (?, ?, ?)", sample_data)

# Commit the changes to the database
conn.commit()


# Your SQL query (remains the same)
query = """
SELECT
    STRFTIME('%Y', "Order Date") AS year,
    STRFTIME('%m', "Order Date") AS month,
    SUM(turnover) AS total_revenue,
    COUNT(DISTINCT "Order ID") AS order_volume
FROM
    sales_data
WHERE
    DATE("Order Date") BETWEEN '2019-01-01' AND '2019-12-31'
GROUP BY
    year, month
ORDER BY
    year, month;
"""

# Execute the query
cursor.execute(query)

# Fetch the results
results = cursor.fetchall()

# Print the results
for row in results:
    print(row)

# Close the connection
conn.close()
```
